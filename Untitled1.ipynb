```python
import nltk
import pandas as pd
import re
import string

from nltk.corpus import movie_reviews, stopwords
from nltk.stem import PorterStemmer

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

nltk.download('movie_reviews')
nltk.download('stopwords')

texts = []
labels = []

for category in movie_reviews.categories():
    for file_id in movie_reviews.fileids(category):
        texts.append(movie_reviews.raw(file_id))
        labels.append(category)

df = pd.DataFrame({
    'text': texts,
    'label': labels
})

print(df.shape)
print(df.head())
print(df['label'].value_counts())
```

```python
stop_words = set(stopwords.words('english'))
stemmer = PorterStemmer()

def clean_text(text):
    text = text.lower()
    text = re.sub(r'\d+', ' ', text)
    text = text.translate(str.maketrans('', '', string.punctuation))
    text = re.sub(r'\s+', ' ', text).strip()

    tokens = text.split()
    tokens = [
        stemmer.stem(word)
        for word in tokens
        if word not in stop_words and len(word) > 1
    ]

    return ' '.join(tokens)

df['clean_text'] = df['text'].apply(clean_text)

print(df[['text', 'clean_text']].head(3))
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    df['clean_text'],
    df['label'],
    test_size=0.2,
    random_state=42,
    stratify=df['label']
)

print("Обучающих текстов:", len(X_train))
print("Тестовых текстов:", len(X_test))
```

```python
vectorizer = TfidfVectorizer(
    min_df=2,
    max_df=0.90,
    max_features=10000
)

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

print("Размер обучающей матрицы:", X_train_tfidf.shape)
print("Размер тестовой матрицы:", X_test_tfidf.shape)
print("Количество слов в словаре:", len(vectorizer.vocabulary_))
```

```python
model = LogisticRegression(max_iter=1000)
model.fit(X_train_tfidf, y_train)

y_pred = model.predict(X_test_tfidf)

print("Accuracy:", round(accuracy_score(y_test, y_pred), 4))
print()
print(classification_report(y_test, y_pred))
print("Матрица ошибок:")
print(confusion_matrix(y_test, y_pred))
```

```python
results = pd.DataFrame({
    'text': X_test.reset_index(drop=True),
    'true_label': y_test.reset_index(drop=True),
    'predicted_label': y_pred
})

errors = results[
    results['true_label'] != results['predicted_label']
].head(5)

for i, row in enumerate(errors.itertuples(index=False), 1):
    print(f"\nОШИБКА {i}")
    print("Правильный класс:", row.true_label)
    print("Предсказание модели:", row.predicted_label)
    print("Текст:", row.text[:1000])
    print("-" * 80)
```

```python
error_analysis = errors.copy()

error_analysis['possible_reason'] = [
    'В тексте присутствует сарказм',
    'Положительные и отрицательные слова смешаны',
    'Не учитывается контекст предложения',
    'Стемминг изменил форму значимых слов',
    'Недостаточно информативных признаков TF-IDF'
][:len(error_analysis)]

display(error_analysis[
    ['true_label', 'predicted_label', 'possible_reason', 'text']
])
```
